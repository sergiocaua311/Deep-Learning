# Comparação de arquiteturas no ESC-50

Este notebook lê os artefatos do pipeline. Ele não repete treinamentos.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
SCREENING = ROOT / 'artifacts/screening/summary.csv'
CROSS_VALIDATION = ROOT / 'artifacts/cross_validation/summary.csv'
```

## Triagem: fold 4 para seleção e fold 5 para teste cego

```python
if SCREENING.exists():
    screening = pd.read_csv(SCREENING)
    display(screening.sort_values('validation_f1_macro', ascending=False))
else:
    print('Execute: python -m esc50 screen')
```

## Validação cruzada oficial dos dois finalistas

```python
if CROSS_VALIDATION.exists():
    cv = pd.read_csv(CROSS_VALIDATION)
    display(cv)
    ax = cv.plot.bar(x='model_family', y=['f1_macro_mean', 'accuracy_mean'], ylim=(0, 1), figsize=(8, 4))
    ax.set_title('Desempenho médio nos cinco folds')
    ax.set_ylabel('Métrica')
    plt.tight_layout()
else:
    print('Execute a validação cruzada depois da triagem.')
```

## Interpretação

Compare F1 macro, variância entre folds, latência e tamanho. O AST histórico (92,75%) não deve ser colocado na mesma coluna do teste cego porque seu fold 5 participou da escolha do checkpoint.
